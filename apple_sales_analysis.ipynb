## Apple products by Products 

```python
import matplotlib.pyplot as plt

top_products = df["Product Name"].value_counts().head(10)

top_products.plot(kind="bar", figsize=(10, 6))

plt.title("Top 10 Apple Products")
plt.xlabel("Product Name")
plt.ylabel("Number of Products")
plt.xticks(rotation=45, ha="right")
plt.show()
```

## Sale Price distribution

```python
plt.hist(df["Sale Price"], bins = 10)

plt.title("distribution of sales price")
plt.xlabel("sale price")
plt.ylabel("number of products")
plt.show()
```

## MRP vs Sale Price

```python
plt.scatter(df["Mrp"], df["Sale Price"])

plt.title("Mrp vs Sale Price")
plt.xlabel("Mrp")
plt.ylabel("Sale Price")
plt.show()
```

## Discount Percentage

```python
plt.hist(df["Discount Percentage"], bins =10)

plt.title("distribution of discount percentage")
plt.xlabel("discount percentage")
plt.ylabel("number of products")
plt.show()
```

## Product ratings

```python
plt.hist(df["Star Rating"] , bins = 10)

plt.title("distribution of star rating")
plt.xlabel("rating")
plt.ylabel("number of products")
plt.show()
```

```python
df.columns
```

## rating vs review

```python
plt.scatter(df["Number Of Ratings"], df["Number Of Reviews"])

plt.title("rating vs review")
plt.xlabel("Number of Ratings")
plt.ylabel("Number of Reviews")
plt.show()
```

## Top 10 products by number of ratings

```python
top_products = df.nlargest(10,"Number Of Ratings")

plt.bar(top_products["Product Name"], top_products["Number Of Ratings"])

plt.title("top 10 products by number of ratings")
plt.xlabel("products")
plt.ylabel("number of ratings")
plt.xticks(rotation = 90)
plt.show()
```

## Top 10 products by reviews

```python
top_reviews = df.nlargest(10, "Number Of Reviews")

plt.bar(top_reviews["Product Name"], top_reviews["Number Of Reviews"])

plt.title("Top 10 Products by Number of Reviews")
plt.xlabel("Product")
plt.ylabel("Number of Reviews")
plt.xticks(rotation=90)
plt.show()
```

## Correlation heatmap

```python
plt.figure(figsize =(10,6))

sns.heatmap(
    df.select_dtypes(include = "number").corr(),
    annot = True,
    cmap = "coolwarm"
)
plt.title("correlation heatmap")
plt.show()
```
